```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sdv.metadata import Metadata
from sdv.single_table import CTGANSynthesizer

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

from catboost import CatBoostClassifier
from xgboost import XGBClassifier

from scipy.stats import ks_2samp

import warnings
warnings.filterwarnings("ignore")

print("All libraries imported successfully!")
```

```python
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]

df = pd.read_csv("/content/adult (1).csv")

print("Dataset loaded successfully!")
print("File:", filename)
print("Dataset shape:", df.shape)

display(df.head(10))
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
)

print("Columns:")
print(df.columns.tolist())
```

```python
# Remove spaces from string values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype(str).str.strip()

# Convert ? to missing
df = df.replace("?", np.nan)

# Remove missing rows
df = df.dropna().reset_index(drop=True)

print("After cleaning:")
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("Income distribution:")

print(
    df["income"].value_counts()
)

print("\nPercentage:")
print(
    df["income"].value_counts(normalize=True) * 100
)
```

```python
target = "income"

feature_columns = [
    col for col in df.columns
    if col != target
]

numeric_columns = df[
    feature_columns
].select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = df[
    feature_columns
].select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Numerical columns:")
print(numeric_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

```python
private_df = df.copy()

np.random.seed(42)

# Controlled privacy noise
private_df["age"] = (
    private_df["age"] +
    np.random.laplace(
        loc=0,
        scale=0.5,
        size=len(private_df)
    )
).clip(17, 90)

private_df["hours.per.week"] = (
    private_df["hours.per.week"] +
    np.random.laplace(
        loc=0,
        scale=0.5,
        size=len(private_df)
    )
).clip(1, 99)

print("Privacy-preserving preprocessing completed!")

display(private_df.head(10))
```

```python
metadata = Metadata.detect_from_dataframe(
    data=private_df,
    table_name="adult_privacy_data"
)

metadata.validate()

print("Metadata created successfully!")
```

```python
synthesizer = CTGANSynthesizer(
    metadata,
    epochs=300,
    verbose=True
)

print("Starting CTGAN training...")

synthesizer.fit(private_df)

print("CTGAN training completed!")
```

```python
synthetic_df = synthesizer.sample(
    num_rows=len(private_df)
)

print("Synthetic data generated!")
print("Synthetic shape:", synthetic_df.shape)

display(synthetic_df.head(10))
```

```python
print("Original dataset:")
print(df.shape)

print("\nSynthetic dataset:")
print(synthetic_df.shape)

print("\nSynthetic missing values:")
print(synthetic_df.isnull().sum())
```

```python
synthetic_df.to_csv(
    "privacy_preserving_synthetic_adult.csv",
    index=False
)

print("Synthetic dataset saved successfully!")
```

```python
X_train = synthetic_df.drop(
    columns=[target]
).copy()

y_train = synthetic_df[target].astype(str)

X_test = df.drop(
    columns=[target]
).copy()

y_test = df[target].astype(str)

# Make categorical columns strings
for col in categorical_columns:

    X_train[col] = X_train[col].astype(str)
    X_test[col] = X_test[col].astype(str)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
cat_model = CatBoostClassifier(
    iterations=800,
    depth=8,
    learning_rate=0.04,
    loss_function="Logloss",
    eval_metric="Accuracy",
    random_seed=42,
    verbose=100,
    l2_leaf_reg=5,
    random_strength=1,
    border_count=128,
    thread_count=-1
)

print("Training CatBoost...")

cat_model.fit(
    X_train,
    y_train,
    cat_features=categorical_columns
)

print("CatBoost training completed!")
```

```python
cat_predictions = cat_model.predict(X_test)

cat_predictions = np.array(
    cat_predictions
).reshape(-1).astype(str)

cat_accuracy = accuracy_score(
    y_test,
    cat_predictions
)

cat_precision = precision_score(
    y_test,
    cat_predictions,
    average="weighted"
)

cat_recall = recall_score(
    y_test,
    cat_predictions,
    average="weighted"
)

cat_f1 = f1_score(
    y_test,
    cat_predictions,
    average="weighted"
)

print("\n======================================")
print("       CATBOOST RESULTS")
print("======================================")

print(f"Accuracy  : {cat_accuracy * 100:.2f}%")
print(f"Precision : {cat_precision * 100:.2f}%")
print(f"Recall    : {cat_recall * 100:.2f}%")
print(f"F1-Score  : {cat_f1 * 100:.2f}%")
```

```python
# Convert categorical data to one-hot encoding

combined = pd.concat(
    [
        synthetic_df.drop(columns=[target]),
        df.drop(columns=[target])
    ],
    axis=0,
    ignore_index=True
)

combined_encoded = pd.get_dummies(
    combined,
    columns=categorical_columns,
    drop_first=False
)

X_train_xgb = combined_encoded.iloc[
    :len(synthetic_df)
].copy()

X_test_xgb = combined_encoded.iloc[
    len(synthetic_df):
].copy()

# Encode target
label_encoder = LabelEncoder()

all_y = pd.concat(
    [
        synthetic_df[target].astype(str),
        df[target].astype(str)
    ]
)

label_encoder.fit(all_y)

y_train_xgb = label_encoder.transform(
    synthetic_df[target].astype(str)
)

y_test_xgb = label_encoder.transform(
    df[target].astype(str)
)

print("XGBoost data prepared!")
print("Training shape:", X_train_xgb.shape)
print("Testing shape:", X_test_xgb.shape)
```

```python
xgb_model = XGBClassifier(
    n_estimators=700,
    max_depth=6,
    learning_rate=0.04,
    min_child_weight=2,
    subsample=0.9,
    colsample_bytree=0.9,
    gamma=0,
    reg_alpha=0.05,
    reg_lambda=1.0,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

print("Training XGBoost...")

xgb_model.fit(
    X_train_xgb,
    y_train_xgb
)

print("XGBoost training completed!")
```

```python
xgb_predictions = xgb_model.predict(
    X_test_xgb
)

xgb_accuracy = accuracy_score(
    y_test_xgb,
    xgb_predictions
)

xgb_precision = precision_score(
    y_test_xgb,
    xgb_predictions,
    average="weighted"
)

xgb_recall = recall_score(
    y_test_xgb,
    xgb_predictions,
    average="weighted"
)

xgb_f1 = f1_score(
    y_test_xgb,
    xgb_predictions,
    average="weighted"
)

print("\n======================================")
print("       XGBOOST RESULTS")
print("======================================")

print(f"Accuracy  : {xgb_accuracy * 100:.2f}%")
print(f"Precision : {xgb_precision * 100:.2f}%")
print(f"Recall    : {xgb_recall * 100:.2f}%")
print(f"F1-Score  : {xgb_f1 * 100:.2f}%")
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Previous Random Forest",
        "CatBoost",
        "XGBoost",
        "Reference Project"
    ],

    "Accuracy": [
        79.57,
        cat_accuracy * 100,
        xgb_accuracy * 100,
        81.30
    ]
})

display(comparison)
```

```python
best_accuracy = max(
    cat_accuracy,
    xgb_accuracy
)

if cat_accuracy >= xgb_accuracy:
    best_model_name = "CatBoost"
else:
    best_model_name = "XGBoost"

print("======================================")
print("        BEST MODEL RESULT")
print("======================================")

print("Best Model:", best_model_name)
print(f"Best Accuracy: {best_accuracy * 100:.2f}%")

if best_accuracy * 100 >= 81.6:
    print("\nTARGET ACHIEVED!")
    print("Accuracy is above 81.6%.")
else:
    print("\nTarget not reached yet.")
    print("We need another optimization experiment.")
```

```python
numerical_columns = [
    col for col in numeric_columns
    if col in synthetic_df.columns
]

fidelity_results = []

for col in numerical_columns:

    real_values = pd.to_numeric(
        df[col],
        errors="coerce"
    ).dropna()

    synthetic_values = pd.to_numeric(
        synthetic_df[col],
        errors="coerce"
    ).dropna()

    ks_stat, p_value = ks_2samp(
        real_values,
        synthetic_values
    )

    fidelity_results.append({
        "Feature": col,
        "KS Statistic": ks_stat,
        "P-Value": p_value
    })

fidelity_df = pd.DataFrame(
    fidelity_results
)

display(fidelity_df)
```

```python
real_records = set(
    df.astype(str).apply(
        lambda row: "|".join(row),
        axis=1
    )
)

synthetic_records = set(
    synthetic_df.astype(str).apply(
        lambda row: "|".join(row),
        axis=1
    )
)

matching_records = (
    real_records.intersection(
        synthetic_records
    )
)

duplication_rate = (
    len(matching_records) /
    len(synthetic_records)
) * 100

print("======================================")
print("       PRIVACY EVALUATION")
print("======================================")

print(
    "Exact matching records:",
    len(matching_records)
)

print(
    f"Exact duplication rate: {duplication_rate:.4f}%"
)
```

```python
final_results = pd.DataFrame({
    "Evaluation": [
        "Data Fidelity",
        "ML Accuracy",
        "ML Precision",
        "ML Recall",
        "ML F1-Score",
        "Privacy - Exact Duplication"
    ],

    "Metric": [
        "Average KS Statistic",
        "Best Model Accuracy",
        "Best Model Precision",
        "Best Model Recall",
        "Best Model F1-Score",
        "Exact Match Rate"
    ],

    "Result": [
        f"{fidelity_df['KS Statistic'].mean():.4f}",
        f"{best_accuracy * 100:.2f}%",
        (
            f"{cat_precision * 100:.2f}%"
            if cat_accuracy >= xgb_accuracy
            else f"{xgb_precision * 100:.2f}%"
        ),
        (
            f"{cat_recall * 100:.2f}%"
            if cat_accuracy >= xgb_accuracy
            else f"{xgb_recall * 100:.2f}%"
        ),
        (
            f"{cat_f1 * 100:.2f}%"
            if cat_accuracy >= xgb_accuracy
            else f"{xgb_f1 * 100:.2f}%"
        ),
        f"{duplication_rate:.4f}%"
    ]
})

display(final_results)
```
